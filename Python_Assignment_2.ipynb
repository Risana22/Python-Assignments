**Performance Category Program:**
1. Prompt user for Input. Score range should be from 0 to 10 (both inclusive).
2. Find the performance category based on the input score using following criteria:

a. Above Average: Score greater than 7

b. Average: Score between 4 and 7(both inclusive)

c. Below Average: Score lesser than 4

4. Additional Step: You can give a prompt of your choice to each category.
For eg: If score below average “Need to Improve your performance, consistent
practice will lead to better results”.

```python
print("Performance Category Program")
score=int(input("Enter your score(0-10):"))
if score>7:
 print( "'Above Average':","Excellent! You Nailed it!")
elif score>=4 and score<=7:
  print( "'Average':","Great effort! Keep practising, Practise makes a man Perfect!")
else:
  print( "'Below Average':","Good effort! Nothing is impossible, try hard and crack it!")

```

 **Selection statement if-elif-else is used here For creating a performance category program. In this every time we have to run the program for knowing the category performance of different scores. And added some prompts for for each performance category.**
