**Number Guessing Game**

Problem Statement: Create a Python program that implements a simple number
guessing game using a while loop. The program should make use of control
statements such as else, break, and continue.

**1. Set Up the Game:**

★ Generate a random number between 1 and 10 that the user has to guess.
Import random and use randint function.

**2. Prompt the User:**

★ Ask the user to guess the number.
★ Set a variable attempts to 3,
which represents the maximum number of
guesses allowed.

**3. Implement the Guessing Logic:**

★ Use a while loop to allow the user to keep guessing until they
 get the
correct number or run out of attempts.

★ Provide feedback to the user for each guess:

➢ If the guess is out of the valid range (1 to 10), inform the user.

➢ If the guess is greater than the secret number.

➢ If the guess is lower than the secret number.

➢ If the guess is correct, congratulate the user and end the game.

**4. Control Statements:**

★ Use continue to skip the rest of the loop after informing the user if the
guess is out of range.

★ Use break to exit the loop when the guess is correct.

★ Use else with the while loop to provide a message like "Better luck next
time!" if the user runs out of attempts without guessing the correct number.

```python

import random
secret_number=random.randint(1,10)
attempts=3
while attempts>0:
  guess=int(input("Guess the number between 1 and 10: "))
  print(guess)
  if guess<1 or guess>10:
    print("You are out of the range, Please guess a number between 1 and 10")
    attempts-=1
    continue
  elif guess==secret_number:
    print("Congratulations your guess is absolutely correct!")
    break
  elif guess>secret_number:
    print("Your guess is too high, Try again...!")
  else:
    print("Your guess is too low, Try again...!")
  attempts-=1
else:
    print("Your chances are over...!  'Better Luck next time...!'")

```

### **Number Guessing Game**

**In the above generated number guessing game is created by using while loop along with if-elif-else control statement, while-else, break and continue inside while loop.**

 **The code starts with importing random function and randint function to get a random number for the guessing game.Then input keyword is used for asking the user to guess for participating in the game, and it is stored in the variable named secret_number. Then for giving proper instructions and feedback regarding the numbers guessed by the player, if-elif-else is used inside while loop.**

 **The while loop is started with maximum attempts as 3, then for the running of loop decrement is declared clearly.Input keyword is used inside a variable named 'guess', then certain conditions are given in if and elifs for the feedback and instructions.**

**In between the while loop along with if-elif-else, break is used to stop the loop when the guess is correct and continue is used to skip the guess if the guess is out of the range.**

**Finally if the loop stops due to out of the attempts then using while-else a 'better luck next time' messege is set.**

**Multiplication Table Generator**

Problem Statement:

Create a Python program that generates and prints a
multiplication table (from 1 to 10) for a given number using a for loop and the range
function.
Step wise Instructions:
1. Prompt user for Input. Ask the user to enter a number for which they want to
generate a multiplication table.
2. Generate the Multiplication Table: Use a for loop to iterate through the numbers
1 to 10. In each iteration, calculate the product of the user's number and the
current number from the loop.
3. Display the Multiplication Table:
Print each line of the multiplication table in the format: "number x i = result".

```python
number=int(input("Multiplication table of: "))
for i in range(1,11):
  print(number,"*",i,"=",number*i)
```

**BMI Calculator**


Problem Statement: Create a Python program that calculates the Body Mass Index
(BMI).

Hint: BMI = weight (kg) / [height (m)]2
Instructions:
1. Define a function calculate_bmi(weight, height) that returns the BMI.
2. Prompt the user for their weight (in kg) and height (in meters).
3. Use the function to calculate and display the BMI

```python
def calculate_bmi(weight,height):
  weight=float((input("Enter your weight(in kgs)= ")))
  height=float((input("Enter your height(in meters)= ")))
  BMI=round(weight/(height**2),2)
  print("BMI of the given weight and height = ",BMI)

calculate_bmi(55,1.5)
```

**BMI Calculating function**

**The function calculate_bmi() is created to calculate BMI of the given weight and height. The function created here refers two arguments weight and height, it is instructed to the users that weight should be given in kgs and height should be in meters when they are asked to provide both. Input keyword is used along with float keyword(since kgs and meters will have decimal points) when we provides the prompt for weight and height. Then the expression used here is the formula for calculating BMI. Finally when we calling the function we have to specify with the height and weight along with function name, then the input function will act and user have to provide according to the instructions.**
